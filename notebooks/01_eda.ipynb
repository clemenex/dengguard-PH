```python
import pandas as pd
import matplotlib as plt

from pathlib import Path
import os
```

```python
REPO_ROOT = Path.cwd().resolve().parent
```

```python
df = pd.read_csv(REPO_ROOT / "data" / "denggue-cases.csv")

print(df.head())
print(df.columns)
print(df.isna().sum())
```

```python
df = df.rename(columns={
    "Month": "month_name",
    "Year": "year",
    "Region": "region",
    "Dengue_Cases": "cases",
    "Dengue_Deaths": "deaths"
})
```

```python
month_map = {
    "January": 1, "February": 2, "March": 3, "April": 4,
    "May": 5, "June": 6, "July": 7, "August": 8,
    "September": 9, "October": 10, "November": 11, "December": 12
}
df["month"] = df["month_name"].map(month_map)
```

```python
df["date"] = pd.to_datetime(dict(
    year=df["year"],
    month=df["month"],
    day=1
))

print(df[["date", "region", "cases"]].head())
```

```python
clean_path = REPO_ROOT / "data" / "dengue-cases.csv"
df.to_csv(clean_path, index=False)

print("saved to:", clean_path)
```
